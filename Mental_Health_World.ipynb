## Looking at the data and comparing countries

#### Here are a few countries I compared

```python
countries = ["United States", "United Kingdom", "Australia", "Greece"]

mental_health_rates = [18.0, 22.0, 21.7, 19.8]

print(countries)
print(mental_health_rates)
```

#### I decided to calculate the average of the countries I selected 

```python
average_rate = sum(mental_health_rates) / len(mental_health_rates)

print("Average rate:", average_rate, "%")
```

#### I decided to see which country in my sample had the highest percentage

```python
highest_rate = max(mental_health_rates)
highest_country = countries[mental_health_rates.index(highest_rate)]

print("Highest rate in my sample:", highest_country)
print("Percentage:", highest_rate, "%")
```

## What does the data tell us?

### The countries in my sample had different percentages of people with mental health disorders. The average of my selected countries was 20.375%. This shows that mental health disorders affect a high number of populations even though the percentages vary between the countries.
